```python
# This script creates 99 text files named testfile_1.txt to testfile_99.txt
# Each file contains a line indicating its number.
import os
# and then deletes them immediately after creation.

for i in range(1, 100):
    #print(os.getcwd())  # Print the current working directory\
    #print(f"Creating file: testfile_{i}.txt")
    
    filename = f"testfile_{i}.txt"
    with open(filename, 'w') as f:
        f.write(f"This is test file number {i}\n")
    print(f"Created {filename}")

    os.remove(filename)
    print(f"Deleted {filename}")
# Note: This script will create and delete files in the current working directory.
# End of script

```
